```python
selected_category = "Skuespil"  # Eksakt verdi i genre-kolonnen

# Les Excel-filen
meta_df = pd.read_excel(metadata_sheet, header=0, index_col=False)

# Filtrer: eksakt match i genre + urn finnes
filtered_df = meta_df[
    (meta_df["genre"] == selected_category) &
    (meta_df["urn"].notna())
]

# Lag liste
urns = filtered_df["urn"].tolist()

# Vis noen
print(f"Fant {len(urns)} URN-er:")
print(urns[:5])
```

```python
# Lag korpus
corpus = dh.Corpus()
corpus.extend_from_identifiers(identifiers=urns)
```

```python
corpus
```

```python
# Frekvenser per dokument
counts = Counts(corpus)

# Hent underliggende DataFrame og summer over alle kolonner (bøker)
total_freq = counts.frame.sum(axis=1).sort_values(ascending=False)

# Vis de 20 mest brukte ordene
print(total_freq.head(20))
```

```python

# === Funksjon for å hente URN-er for én sjanger ===
def get_urns_for_genre(df, genre):
    genre_filtered = df[
        (df["genre"].str.strip().str.casefold() == genre.casefold()) &
        (df["urn"].notna()) &
        (df["urn"].str.startswith("URN"))
    ]
    return genre_filtered["urn"].tolist()

# === Funksjon for toppord med frekvens ===
def get_top_words_with_freq(urns, start=50, end=70):
    corpus = dh.Corpus()
    corpus.extend_from_identifiers(identifiers=urns)
    counts = Counts(corpus)
    freq_series = counts.frame.sum(axis=1).sort_values(ascending=False)
    selected = freq_series.iloc[start:end]
    return [f"{word} ({count})" for word, count in selected.items()]

# === Hent metadata ===
meta_df = pd.read_excel("Sjangre_kategorisert_010625.xlsx")

# === Sjangre og intervall ===
genre1 = "Lystspil"
genre2 = "Sørgespil"
start_rank = 50
end_rank = 70

# === Få URNs og toppord ===
urns1 = get_urns_for_genre(meta_df, genre1)
urns2 = get_urns_for_genre(meta_df, genre2)

top_words1 = get_top_words_with_freq(urns1, start_rank, end_rank)
top_words2 = get_top_words_with_freq(urns2, start_rank, end_rank)

# === Kombinér i én DataFrame ===
comparison_df = pd.DataFrame({
    genre1: top_words1,
    genre2: top_words2
})

# === Vis ===
print(comparison_df)
```
